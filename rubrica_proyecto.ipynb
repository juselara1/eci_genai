Rúbrica del proyecto

En este proyecto aplican un modelo generativo o un agente basado en
LLMs a un problema práctico. El tema es libre, pero la solución debe
resolver un problema real y concreto con una implementación funcional.

Para la entrega deben:

1. Grabar un video de 5 minutos mostrando el funcionamiento de su
   implementación, dejarlo público y únicamente adjuntar el enlace
   (en Drive o en YouTube).
2. Adjuntar el código de la implementación.

**Enlace de entrega:** https://forms.gle/iR28e8DfWrxkuz2A6

Criterios de evaluación

| Criterio | Puntaje |
|---|---|
| Problema práctico y justificación | 15 |
| Aplicación de GenAI | 25 |
| Implementación técnica | 25 |
| Video | 20 |
| Resultados y evaluación | 10 |
| Documentación y reproducibilidad | 5 |

**Total: 100 puntos**

## Problema práctico y justificación (15 puntos)

| Nivel | Puntaje |
|---|---|
| El problema es real, concreto y la justificación explica claramente por qué GenAI aporta valor frente a una solución tradicional. | 12-15 |
| El problema es válido, pero la justificación es superficial o la comparación con alternativas es débil. | 8-11 |
| El problema es demasiado abstracto o el uso de GenAI parece innecesario. | 0-7 |

## Aplicación de GenAI (25 puntos)

| Nivel | Puntaje |
|---|---|
| Desarrollan un LLM o agente con decisiones técnicas justificadas: arquitectura, datos, prompting, tools, retrieval o fine-tuning según el problema. | 20-25 |
| Usan GenAI de forma correcta pero con poco diseño propio; el sistema se apoya casi por completo en un modelo o API existente sin adaptación. | 12-19 |
| La aplicación de GenAI es marginal, decorativa o no corresponde al problema. | 0-11 |

## Implementación técnica (25 puntos)

| Nivel | Puntaje |
|---|---|
| La implementación es funcional, robusta y maneja entradas inválidas, límites del modelo y errores de ejecución. El código es claro y modular. | 20-25 |
| La implementación funciona en el caso principal, pero falla o no define comportamiento en casos comunes. | 12-19 |
| La implementación es inestable, incompleta o difícil de mantener. | 0-11 |

## Video (20 puntos)

| Nivel | Puntaje |
|---|---|
| El video es público, dura máximo 5 minutos y muestra claramente el flujo completo: entrada, procesamiento con GenAI, salida y casos representativos. | 16-20 |
| El video es público, pero no muestra casos importantes o la explicación es poco clara. | 10-15 |
| El video no es público, excede 5 minutos o no permite evaluar la implementación. | 0-9 |

## Resultados y evaluación (10 puntos)

| Nivel | Puntaje |
|---|---|
| Presentan métricas o evaluación cualitativa sistemática y discuten limitaciones con evidencia del sistema real. | 8-10 |
| Presentan resultados, pero la evaluación es anecdótica o no cubre las capacidades clave del sistema. | 5-7 |
| No presentan resultados verificables o no discuten limitaciones. | 0-4 |

## Documentación y reproducibilidad (5 puntos)

| Nivel | Puntaje |
|---|---|
| El código está organizado, documentado y permite reproducir el demo con instrucciones de instalación y uso. | 4-5 |
| La documentación existe pero es incompleta o la ejecución requiere pasos no explicados. | 2-3 |
| No hay instrucciones suficientes para reproducir la solución. | 0-1 |
