```python
from transformers import pipeline
```

# Clasificacion

```python
model = pipeline(task="text-classification", model="cardiffnlp/twitter-roberta-base-sentiment")
```

```python
model("I'm so sad today")
```

```python
model("I was happy yesterday because it was my birthday")
```

# Token Classification

```python
model = pipeline(task="token-classification", model="dslim/bert-base-NER")
```

```python
model("I'm in Colombia with my friend Jose")
```

# Zero-Shot

```python
model = pipeline(task="zero-shot-classification", model="facebook/bart-large-mnli")
```

```python
model("I like topology theory", ["physics", "math", "chemistry", "sports"])
```

```python
model("I predicted the movement of the pendulum", ["physics", "math", "chemistry", "sports"])
```

```python
model("The water is composed by hydrogen and oxygen", ["physics", "math", "chemistry", "sports"])
```

```python
model("Hugo Calderano is the best table tennis player from Brazil", ["physics", "math", "chemistry", "sports"])
```

# Text Generation

# Summarization

```python
import cactus

@cactus.tool
def get_weather(city: str):
    "Get the current weather for a city."
    return {"city": city, "temp_c": 27, "sky": "clear"}

with cactus.Cactus(tools=[get_weather]) as agent:
    print(agent.run("What's the weather in Lagos?")["results"])
```

```python
with cactus.Cactus() as agent:
    print(agent.run("Hello how are you")["results"])
```
