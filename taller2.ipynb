## 3. Modelo CBOW

En CBOW, dadas $2r$ palabras de contexto
$\{w_{i-r}, \dots, w_{i+r}\}$ alrededor de la palabra central $w_i$,
el modelo busca maximizar $P(w_i \mid \text{contexto})$.

La arquitectura es:

1. Una capa `nn.Embedding` mapea cada palabra de contexto a un vector
   de tamaño `embedding_dim`.
2. Promediamos los vectores de contexto.
3. Una capa lineal (`nn.Linear`) proyecta el vector promediado sobre
   el vocabulario y produce logits.
4. Con `nn.CrossEntropyLoss` entrenamos con el índice de la palabra
   central como objetivo.
